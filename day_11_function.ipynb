```python
# write function to sort the list
def sort_list(values):
    return sorted(values)

numbers = [6, 5, 8, 9 ,100,1,45,80]

sort_list(numbers)
```

```python
# write a function to check if  data contains duplicate or not (set doesn't take duplcate)

# [1,2,3,4] - True no duplicates
#[1,2,3,4,4] - False have duplicates       
#  interview question google ,oracle


def check_duplicates(data):
    return len(data) == len(set(data))


print(check_duplicates([1, 2, 3,4]))

print(check_duplicates([1, 2, 4, 4]))     


```

```python
# 13. Create a function that returns the grade based on marks.
# * Marks ≥ 80 → A
# * Marks ≥ 60 → B
# * Marks ≥ 40 → C
# * Marks < 40 → Fail


def marks(a):
    if a >= 80:
        print("A")
        
        
    elif a>=60:
        print("B")   
        
        
    elif a>=40:
        print("C")
        
    elif a<=40:
            print("fail")
                
marks(50)             
marks(10) 
marks(90)  
marks(75)                    
             
    
```
