```python
# 7. Take three numbers and find the largest among them.

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

if (a >= b) and (a >= c):
    largest = a
elif (b >= a) and (b >= c):
    largest = b
else:
    largest = c

print(f"The largest number is: {largest}")
```

```python
# 10. Take marks as input and print the grade:
# - 90 and above → A
# - 75 to 89 → B
# - 50 to 74 → C
# - Below 50 → Fail

marks = float(input("Enter marks: "))

if marks >= 90:
    print("A")
elif 75 <= marks <= 89:
    print("B")
elif 50 <= marks <= 74:
    print("C")
else:
    print("Fail") 
```

```python
# 10. Take a shopping amount and apply discount:
# - Above 5000 → 20% discount
# - 2000–5000 → 10% discount
# - Below 2000 → No discount




amt = int(input("enter the total shopping amount"))

if amt > 5000:
   print ("20% Discount applied",amt )
   dis = amt *20/100
   print("discounted amount is",dis)
   print('after discount final amt is', amt-dis)

elif amt > 2000 :
    print(" 10 percent Discount apllied")
    dis = amt *10/100
    print("discounted amount is",dis)
    print('after discount final amt is', amt-dis)

  

else :
    print("sorry ! no discount applied")    
```
