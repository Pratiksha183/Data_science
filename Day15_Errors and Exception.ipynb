# Errors and Exceptions in Python

## What are Errors and Exceptions?
- **Errors**: Problems that stop code execution.
  - *Syntax Errors*: Mistakes in code structure (e.g., missing colon).
  - *Runtime Errors*: Occur while the program runs (e.g., division by zero).
  
- **Exceptions**: Special runtime errors that Python can *catch and handle*.

- What are Exceptions?
  - Exceptions are runtime errors that occur when a program encounters something unexpected.Instead of stopping execution abruptly, Python provides a way to detect and handle these exceptions gracefully.
# Why are Exceptions Needed?
- Prevent crashes: Programs don’t terminate suddenly when errors occur.
- Improve user experience: Users get meaningful error messages instead of cryptic failures.
- Aid debugging: Exceptions identify the exact cause of failure.
- Support robustness: Applications remain reliable even when unexpected inputs or conditions arise.
- Encourage safe design: Developers can anticipate and handle potential problems in advance.

## 10 Common Exception Types in Python

 - ValueError → Raised when a function receives an argument of the right type but inappropriate value.
- TypeError → Occurs when an operation is applied to an object of an invalid type.
- IndexError → Triggered when accessing a list or tuple index that is out of range. 
- KeyError → Raised when trying to access a dictionary key that does not exist.
- ZeroDivisionError → Occurs when division or modulo by zero is attempted.
- FileNotFoundError → Raised when a file or directory is requested but not found.
- ImportError → Triggered when an import statement fails to find or load a module.
- AttributeError → Occurs when an invalid attribute reference is made on an object.
- OSError → Raised for operating system-related errors (like file access issues).
- RuntimeError → Raised when an error doesn’t fall into other specific categories. 

## Handling Errors with Try-Except
- `try`: Place risky code here.
- `except`: Runs if an error occurs.
- Prevents program crashes and allows graceful recovery.
- Multiple `except` blocks can handle different error types.

## Raising Exceptions
- Use `raise` to **manually trigger an error**.
- Helps enforce rules or validate inputs.
- Common in APIs, data validation, and custom logic.

## Using Finally for Cleanup
- `finally`: Always executes, whether error occurs or not.
- Ideal for cleanup tasks:
  - Closing files
  - Releasing resources
  - Disconnecting from databases
- Ensures program leaves no loose ends.

## Why This Matters
- Makes programs **robust and reliable**.
- Improves **user experience** by avoiding crashes.
- Essential for **production-level code** where stability is critical.

# Syntax
- Python provides four main keywords for handling exceptions: try, except, else and finally each plays a unique role. Let's see syntax:

try:
      # Code 
except SomeException:
      # Code 
else:
     # Code 
finally:
    # Code 

- try: Runs the risky code that might cause an error.
- except: Catches and handles the error if one occurs.
- else: Executes only if no exception occurs in try.
- finally: Runs regardless of what happens useful for cleanup tasks like closing files.

```python
try:
    n = 0
    res = 100 / n
    
except ZeroDivisionError :
    print("You can't divide by zero!")
    
except ValueError:
    print("Enter a valid number!")
    
else:
    print("Result is", res)
    
finally:
    print("Execution complete.")
```
