# Indexing
- Every item in a list has a position number called an index. Indexing allows us to access a single item from the list using its position.
- Positive indexing starts from 0 for the first item, 1 for the second item, and so on.
- Negative indexing starts from -1 for the last item, -2 for the second last item, and so on.
- Negative indexing is useful when we want to access items from the end of the list without knowing its exact length.

```python
# Indexing

print(x[-4])

print(x[4:8])

print(x[-4:-1])
```

# list : Methods
- append() → Adds an item to the end of the list.
- insert() → Inserts an item at a specified position.
- extend() → Adds all elements of another list (or iterable) to the current list.
- remove() → Removes the first occurrence of a specified value.
- pop() → Removes and returns an item at a given index (default: last item).
- clear() → Removes and returns an item at a given index (default: last item).
- index() → Returns the index of the first occurrence of a specified value.
- count() → Returns the number of times a value appears in the list.
- sort() → Sorts the list in ascending order (can take a key or reverse flag).
- reverse() → Reverses the order of the list.
- copy() → Returns a shallow copy of the list.

```python
# Triangle pattern

for i in range (1,6):
    print("*" * i)
```

```python
# pyrimid pattern
rows = 5

for i in range(1, rows + 1):
    spaces = " " * (rows - i)
    stars = "* " * i
    print(spaces + stars)
```

```python
#Dimond pattern

n = 5

for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)

```

```python
# Hollow square
n = 5

for row in range(n):
    for col in range(n):
        if row == 0 or row == n - 1 or col == 0 or col == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# holow pyrimid
n = 5

for i in range(1, n + 1):
    print(" " * (n - i), end="")

    if i == 1:
        print("*")
    elif i == n:
        print("* " * n)
    else:
        print("*" + " " * (2 * i - 3) + "*")
```
