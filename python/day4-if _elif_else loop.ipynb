```python
print("*********** MINI CALCULATOR ***********")

no1 = int(input("Enter first number: "))
no2 = int(input("Enter second number: "))
op = input("Enter operator (+, -, /, *): ")

if op == "+":
    print("Addition is:", no1 + no2)

elif op == "-":
    print("Subtraction is:", no1 - no2)

elif op == "/":
    print("Division is:", no1 / no2)

elif op == "*":
    print("Multiplication is:", no1 * no2)

else:
    print("Put a valid operator")
```

```python

#Write a program to check if a character is:
#vowel → a, e, i, o, u
#else → consonant

character = input("Enter an alphabet: ")

if character in ("a", "e", "i", "o", "u"):
    print("Vowel")
else:
    print("Consonant")
```

```python
#Ask the user for temperature and print:
#temp >= 40 → Very hot
#temp >= 30 → Hot
#temp >= 20 → Normal
#else → Cold"""


temperature =  int(input("Enter value"))

if temperature >= 40:
    print("very hot")
elif temperature >= 30:
    print("hot")
elif temperature >= 20:
    print("normal")
else:
    print("cold")
```
