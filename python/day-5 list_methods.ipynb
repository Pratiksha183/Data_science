```python
# Triangle pattern

for i in range (1,6):
    print("*" * i)
```

```python
# pyrimid pattern
rows = 5

for i in range(1, rows + 1):
    spaces = " " * (rows - i)
    stars = "* " * i
    print(spaces + stars)
```

```python
#Dimond pattern

n = 5

for i in range(1, n + 1):
    print(" " * (n - i) + "* " * i)

for i in range(n - 1, 0, -1):
    print(" " * (n - i) + "* " * i)

```

```python
# Hollow square
n = 5

for row in range(n):
    for col in range(n):
        if row == 0 or row == n - 1 or col == 0 or col == n - 1:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
# holow pyrimid
n = 5

for i in range(1, n + 1):
    print(" " * (n - i), end="")

    if i == 1:
        print("*")
    elif i == n:
        print("* " * n)
    else:
        print("*" + " " * (2 * i - 3) + "*")
```
