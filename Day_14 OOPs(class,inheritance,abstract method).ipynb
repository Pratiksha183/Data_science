# Object-Oriented Programming (OOP) in Python

## What is OOP?
Object-Oriented Programming (OOP) is a programming paradigm that organizes software design around **objects**.  
Objects represent entities that combine both **data** (attributes) and **behavior** (methods).  
Instead of writing code as a sequence of instructions, OOP models problems by defining classes and creating objects from them.

## How does OOP work?
- A **class** acts as a blueprint that defines attributes and methods.  
- An **object** is an instance created from a class.  
- Each object can store its own data and perform actions defined by the class.  
- This approach makes programs modular, reusable, and easier to maintain.

## Why OOP?
- Provides **modularity**: code is organized into logical units.  
- Encourages **reusability**: classes and methods can be reused across projects.  
- Improves **maintainability**: changes in one part of the code are less likely to break others.  
- Models **real-world problems** naturally by representing entities as objects.

---

## Where is OOP used in the Real World?
- **Banking Systems**: Accounts, customers, and transactions are modeled as objects.  
- **E-commerce Platforms**: Products, shopping carts, and orders are represented as classes and objects.  
- **Game Development**: Characters, weapons, and environments are objects with attributes and behaviors.  
- **Healthcare Applications**: Patients, doctors, and appointments are structured as objects.  
- **Social Media Apps**: Users, posts, and comments are modeled using OOP principles.

---

## OOP in Projects
When building projects, OOP helps in:
- **Structuring code**: Large projects are broken into classes for clarity.  
- **Team collaboration**: Different developers can work on different classes without conflict.  
- **Scalability**: Adding new features becomes easier by extending existing classes.  
- **Real-world modeling**: Projects like a Library Management System, Student Database, or Inventory Tracker use OOP to represent entities and their interactions.

---

## Main Components of OOP in Python
- **Class** → Blueprint that defines attributes and methods.  
- **Object** → Instance created from a class.  
- **Attributes** → Variables inside a class that hold data about the object.  
- **Methods** → Functions inside a class that define the behavior of the object.  
- **Inheritance** → Mechanism to create new classes from existing ones, reusing functionality.  
- **Encapsulation** → Bundling data and methods together, with controlled access.  
- **Polymorphism** → Same method name used in different classes with different behaviors.  
- **Abstraction** → Hiding implementation details and exposing only essential features.

# Inheritance
- 1. single  :- single parent single class
- 2. multiple :- two or more parent class and 1 child class
- 3. multilevel :- chain of parent-child-child property usability
- 4. hierarchical :- Multiple independent child classes all inherit from the same single parent class
- 5. hybrid:- combination of type of inheritances


```python
# 1.single inheritance


class parent:
    def teach(self):
        print("This is parent class")

class Child(parent): # inherited
    def play(self):   
        print("this is child class")

c1 = Child()
c1.teach()
c1.play()
```

```python
# 2. Multiple inheritance


class Father():
    def A(self):
        print('this is a father class')



class Mother:
    def B(self):
        print("this is mother class")   


class Child(Father,Mother):
    def C(self):
        print("this is child class")      



OBJ = Child()
OBJ.A()
OBJ.B()
OBJ.C()                 
```

```python
# 3.Multilevel inheritance


class GrandFather():
    def A(self):
        print('this is a Grand-father class')



class Father(GrandFather):
    def B(self):
        print("this is fatherr class")   


class Child(Father):
    def C(self):
        print("this is child class")    


class Son(Child):  
    def D(self):
        print("this is son class")        



OBJ = Son()
OBJ.A()
OBJ.B()
OBJ.C() 
OBJ.D() 



```

```python
#4. Heirarchical inheritance


class Parent:
    def data(self):
        print("this is parent class")



class Child1(Parent):
    def X(self):
        print("this is child no 1")



class Child2(Parent):
    def Y(self):
        print("this is child no 2")


c1 = Child1()
c2 = Child2()



c1.data()
c1.X()


c2.data()
c2.Y()
```

```python
# 5.Hybrid ineritance : combination of 2


class School:
    def info (self):
        print("this is a school class")

class Student(School):
    def data(self):                              # heirarchical
        print("this is a student data")

class Teacher(School):
    def teach(self):
        print("this is a teacher data")

class Assistant(Student,Teacher):                     # multilevel
    def guide(self):
        print("this is a student & Teachers data")



A1 =Assistant()
A1.data()
A1.teach()
A1.guide()        
```

# Inheritance in Object-Oriented Programming

## What is Inheritance?
Inheritance is a fundamental concept in OOP where a **new class (child or subclass)** derives properties and behaviors from an **existing class (parent or superclass)**.  
It allows the child class to reuse, extend, or modify the functionality defined in the parent class.

## How does Inheritance work?
- A parent class defines common attributes and methods.  
- A child class inherits these features automatically.  
- The child class can add new attributes or methods, or override existing ones.  
- This creates a hierarchy where classes are connected through shared functionality.

## Why is Inheritance used?
- **Code Reusability**: Avoids duplication by reusing logic from the parent class.  
- **Extensibility**: Child classes can add new features or customize behavior.  
- **Maintainability**: Updates in the parent class propagate to child classes.  
- **Hierarchy Representation**: Models real-world relationships naturally, such as Vehicle → Car, Bike.

## Real-World Applications
- **Banking Systems**: A general `Account` class can be inherited by `SavingsAccount` and `CurrentAccount`.  
- **E-commerce Platforms**: A `Product` class can be extended into `Electronics`, `Clothing`, or `Books`.  
- **Game Development**: A `Character` class can be inherited by `Player` and `Enemy`.  
- **Education Systems**: A `Person` class can be extended into `Student` and `Teacher`.

## Summary
Inheritance provides a way to:
- Reuse existing code efficiently.  
- Extend functionality without rewriting.  
- Represent natural hierarchies in programming.  
- Build scalable and maintainable software systems.

```python
class vehicle:
    def __init__(self, brand):
        self.brand = brand

    # method
    def info(self):
        return f"Brand name is {self.brand}"

class Car(vehicle):
    def __init__(self, brand, model):
        # call parent constructor
        # super() → used to call the parent class constructor or methods
        super().__init__(brand)
        self.model = model

    # method
    def info(self):
        return f"Brand is {self.brand} and model is {self.model}"

    # method 
    # Parent class methods can call in child class but child class methods can't call in parent class
    def car_info(self):
        return f'brand is {self.brand} and model is {self.model}'
```

# Abstraction in OOP (Deep Explanation)

## Core Idea
Abstraction is about **hiding the "how" and showing only the "what"**.  
It means: when you use something, you don’t care about its internal complexity — you only care about the essential features that matter to you.

Think of it as a **contract**:  
- The parent class (abstract class) says: "Every child must have this feature."  
- The child class then defines *how* that feature actually works.

---

## Why It Matters
- **Reduces Complexity**: You don’t need to know the internal steps.  
- **Enforces Rules**: Child classes must follow the blueprint.  
- **Focus on Essentials**: You interact with the system at a high level.  
- **Flexibility**: Different child classes can implement the same abstract method in their own way.

---

## Real-World Analogy
- **ATM Machine**:  
  - Abstract idea: "Withdraw money."  
  - You don’t care how the bank server checks balance, deducts money, or updates records.  
  - You only see the interface: insert card, enter PIN, choose amount.  

- **Car Accelerator**:  
  - Abstract idea: "Speed up the car."  
  - You don’t care how fuel mixes with air, how pistons move, or how gears shift.  
  - You only press the pedal and the car accelerates.  

- **Smartphone Unlock**:  
  - Abstract idea: "Unlock the phone."  
  - You don’t care how fingerprint sensors or face recognition algorithms work.  
  - You just touch or look, and the phone unlocks.

---

## Deep Insight
When you define an **abstract method** in a parent class:
- You are saying: *"Every child must implement this method, otherwise it cannot exist as a real object."*  
- This forces consistency across all child classes.  
- Example: If `Device` has an abstract method `power_on()`, then every child (`iPhone`, `iPad`, `MacBook`) must define how it powers on.  

This is why abstraction is called a **blueprint mechanism** — it sets the rules, but leaves the details to the child classes.

---

## Summary
Abstraction is:
- **What it does → visible**  
- **How it does → hidden**  
- A way to enforce rules and reduce complexity.  
- A design principle that makes systems clean, scalable, and easy to use.

```python
from abc import ABC, abstractmethod

# Abstract Parent Class
class Netflix(ABC):
    def __init__(self, company_name):
        self.company_name = company_name
    
    @abstractmethod
    def feature(self):
        pass

# Child Class Streaming
class Streaming(Netflix):
    def __init__(self, company_name, show_name):
        super().__init__(company_name)
        self.show_name = show_name
    
    def feature(self):
        return f"{self.company_name}: Streaming '{self.show_name}' in HD"
    

# Child Class Download
class Download(Netflix):
    def __init__(self, company_name, movie_name):
        super().__init__(company_name)
        self.movie_name = movie_name
    
    def feature(self):
        return f"{self.company_name}: Downloaded '{self.movie_name}' for offline viewing"

# Child Class Recommendation
class Recommendation(Netflix):
    def __init__(self, company_name, user_name):
        super().__init__(company_name)
        self.user_name = user_name
    
    def feature(self):
        return f"{self.company_name}: Personalized recommendations for {self.user_name}"

# Using objects
stream = Streaming("Netflix", "Stranger Things")
download = Download("Netflix", "Inception")
recommend = Recommendation("Netflix", "Rock")

print(stream.feature())      
print(download.feature())
print(recommend.feature())
```

# Encapsulation in Object-Oriented Programming

## What is Encapsulation?
Encapsulation means **wrapping data (variables) and methods (functions) together inside a single unit — the class**.  
It also means **restricting direct access** to some parts of an object, so the internal details are hidden and protected.

In simple words:  
*"Keep the data safe inside the box, and only allow controlled access through doors (methods)."*



## Why Encapsulation?
- Protects data from accidental changes.  
- Provides controlled access using getters and setters.  
- Makes code more secure and organized.  
- Helps maintain consistency in how data is used.



## Simplest Analogy
- **Medicine Capsule**:  
  - The capsule hides the powder inside.  
  - You only consume the capsule, not the raw powder.  
  - This is like a class hiding its data and exposing only safe methods.  

- **Bank Locker**:  
  - Your money is inside the locker (data).  
  - You cannot directly touch it; you need a key or PIN (methods).  
  - Encapsulation ensures only authorized access.  

- **Mobile Phone**:  
  - You press buttons (methods) to make calls or send messages.  
  - You don’t directly interact with the internal circuits (data).  
  - Encapsulation hides complexity and gives you a simple interface.



## Summary
Encapsulation is about:
- **Hiding data inside the class.**  
- **Allowing access only through controlled methods.**  
- **Protecting and organizing code.**

```python
# 3. Polymorphism : Polymorphism means the same method name can behave differently depending on the object.

class Warrior:
    def __init__(self, name):
        self.name = name
    
    def attack(self):
        return f"{self.name} swings a sword!"

class Archer:
    def __init__(self, name):
        self.name = name
    
    def attack(self):
        return f"{self.name} shoots an arrow!"

class Wizard:
    def __init__(self, name):
        self.name = name
    
    def attack(self):
        return f"{self.name} casts a spell!"

# Polymorphism in action  / overriding
players = [
    Warrior("Thor"),
    Archer("Robin"),
    Wizard("Merlin")
]

for player in players:
    print(player.attack())
```

```python
# 4. Abstraction : Abstraction hides implementation details and enforces rules using abstract classes.

from abc import ABC, abstractmethod

# Abstract Parent Class
class Bank(ABC):
    def __init__(self, bank_name):
        self.bank_name = bank_name
    
    @abstractmethod
    def transaction(self, amount):
        pass


# Child Class ATM
class ATM(Bank):
    def __init__(self, bank_name, atm_id):
        super().__init__(bank_name)
        self.atm_id = atm_id
    
    def transaction(self, amount):
        return f"ATM {self.atm_id}: Withdrawn ₹{amount} from {self.bank_name}"


# Child Class OnlineBanking
class OnlineBanking(Bank):
    def __init__(self, bank_name, user_id):
        super().__init__(bank_name)
        self.user_id = user_id
    
    def transaction(self, amount):
        return f"Online Banking: User {self.user_id} transferred ₹{amount} via {self.bank_name}"


# Child Class CashDepositMachine
class CashDepositMachine(Bank):
    def __init__(self, bank_name, machine_id):
        super().__init__(bank_name)
        self.machine_id = machine_id
    
    def transaction(self, amount):
        return f"Machine {self.machine_id}: Deposited ₹{amount} into {self.bank_name}"


# Using objects
atm = ATM("State Bank of Python", "ATM-101")
online = OnlineBanking("State Bank of Python", "User123")
machine = CashDepositMachine("State Bank of Python", "CDM-55")
```

```python
print(atm.transaction(2000))
print(online.transaction(5000))
print(machine.transaction(3000))
```

```python
# encapsulation
class Bank:
    def __init__(self, bank_name, balance):
        # Private attributes
        self.__bank_name = bank_name
        self.__balance = balance

    # Getter methods
    def get_bank_name(self):
        return self.__bank_name
    
    def get_balance(self):
        return self.__balance

    # Public method
    def info(self):
        return f"Bank: {self.__bank_name}, Balance: ₹{self.__balance}"


# Using the encapsulated class
account = Bank("State Bank of Python", 5000)

print(account.info())           # Access via public method
print(account.get_bank_name())  # Access via getter
print(account.get_balance())    # Access via getter
```

## Question
Create a simple **Bank Account System** using Object-Oriented Programming in Python.

### Requirements:
1. Define a class `BankAccount` with:
   - Private attributes: `__owner` and `__balance`.
   - A constructor (`__init__`) to initialize owner name and balance.
   - A method `get_owner()` to return the account owner.
   - A method `get_balance()` to return the current balance.
   - A method `deposit(amount)` to add money to the balance.
   - A method `withdraw(amount)` to subtract money from the balance (only if sufficient funds are available).

2. Create two objects of `BankAccount` and demonstrate:
   - Displaying account info.
   - Depositing money.
   - Withdrawing money.
   - Showing updated balance.

### Hint
Use **encapsulation** by keeping attributes private and accessing them only through getters/setters.


```python
# Bank Account System

class BankAccount:
    def __init__(self, owner , balance):

        # private attribute
        self.__owner = owner
        self.__balance = balance
    
    # Getter method
    def get_owner(self):
        return self.__owner

    def get_balance(self):
        return self.__balance

    # Deposit money
    def deposit(self , amount):
        self.__balance += amount

    # Withdraw Money
    def withdraw(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
        else:
            return f"Balance is low"

    def info(self):
        return f"Account : {self.__owner} and the Balance is {self.__balance}"
     
```

```python
# 4. Create an abstract class Employee with abstract methods work() and salary(). Implement it in Manager class
from abc import ABC, abstractmethod

class Employee(ABC):
    
    @abstractmethod
    def work(self):
        pass
        
    @abstractmethod
    def salary(self):
        pass


class Manager(Employee):
    
    def __init__(self, name, base_salary, bonus):
        self.name = name
        self.base_salary = base_salary
        self.bonus = bonus
        
    def work(self):
        return f"{self.name} is managing projects and teams."
        
    def salary(self):
        return self.base_salary + self.bonus


mgr = Manager("Alice", 80000, 15000)

print(mgr.work())    
print(f"Total Salary: {mgr.salary()} Rs  ") 

```
