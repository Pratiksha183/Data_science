# ARGS* AND KWARGS**

What are *args?
-  *args allows a function to accept so many  number of positional arguments.
- The arguments are collected into a tuple inside the function.
- Useful when you don’t know in advance how many inputs will be passed.

```python
SYNTAX

function_name(*args):
    # function body
```

```python
def show_items(*args):
    print(args)
    
show_items(1,2,6,9,8,10,"radha")    
```

```python
def add(*args):   # accept multiple parameters
    print("all passed elements are →",(args))

          
add(2,4,5)          
```

```python
def sample(**kwargs):    # accept multiple parameter in the form of dictionary
    print(kwargs)
          
sample(name ="pratiksha",age=22)    
```

```python
# 2. Write a function using *args to find the sum of all numbers.

def add(*args): 
    sum = 0
    for i in args:
       sum += i
    return sum   
       
          
add(2,4,5)       
```

```python
#4. Write a function using *args that multiplies all numbers together

def multiply(*args):
    result = 1
    for number in args:
        result *= number
    return result

multiply(10, 2, 3) 
```

# recursive function
- the function that call itself again n again
- same as loop
- has 2 step
1.base case - if part
2.recursive case - else 2 condition

```python
def info(n):
    if n == 0: # base case
        print("Done" )
        
    else :
        print(n)
        info (n-1  ) # recursive call  
info(5)      
```

```python
def factorial(n):
  
   if n == 1: # base case
    return 1
   else:
       return n* factorial(n-1) # recursive no
     
factorial(5)  
```

```python
# loop the key-value pairs

def loop_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(key, value)
```

```python
loop_kwargs(name = "rock" , city = "NYC")
```

```python
def billing(**product):

    
    for key , value in product.items():
        print(key , value)

    amount = 0
    for k , v in product.items():
        amount += v

    return "total amount is " , amount
```

```python
billing(Milk = 20 , Oats = 20, Chocolate = 20 , Sugar = 20)
```

```python
#practice

#Write a function summarize that accept any number of numbers via args and returns a dictionary
#  with count ,total,and average .test it with summerize(10,20,30)

def summarize(*args):
    count = len(args)
    total = sum(args)
    average = total / count if count else 0
    return {"count": count, "total": total, "average": average}

summary = summarize(10, 20, 30)
summary
```

```python
summarize(60,30,566,46)
```
