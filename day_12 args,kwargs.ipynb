What are *args?
- In Python, *args allows a function to accept so many  number of positional arguments.
The arguments are collected into a tuple inside the function.
Useful when you don’t know in advance how many inputs will be passed.

```python
# using args

def add(*args):
    total = 0
    for i in args:
        total += i

    return total
```

def add(a, b) # fixed parameter pass

def add(*nums) # *args passing multiple arguements not fixed

def add(**kwargs) # we will pass arguments like key values pair

```python
# print all keyword argument
def show_kwargs(**kwargs):
    print(kwargs)
```

```python
show_kwargs(city = "pune" , rain = False)
```

```python
def student_info(**record): # pass : same way key and value 
    print(record)
student_info(name = "Rock" , city = "NYC" , age = 20, Grade = "A", sport = "Fighting")
```

```python
# loop the key-value pairs

def loop_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(key, value)
```

```python
loop_kwargs(name = "rock" , city = "NYC")
```

```python
def billing(**product):

    
    for key , value in product.items():
        print(key , value)

    amount = 0
    for k , v in product.items():
        amount += v

    return "total amount is " , amount
```

```python
billing(Milk = 20 , Oats = 20, Chocolate = 20 , Sugar = 20)
```

```python
#practice

#Write a function summarize that accept any number of numbers via args and returns a dictionary
#  with count ,total,and average .test it with summerize(10,20,30)

def summarize(*args):
    count = len(args)
    total = sum(args)
    average = total / count if count else 0
    return {"count": count, "total": total, "average": average}

summary = summarize(10, 20, 30)
summary
```

```python
summarize(60,30,566,46)
```

```python
#Write a function make_tag that uses **kwargs to accept attributes like class, id, and style,
# and returns a dictionary of whatever was passed in Call make_tag(class="btn", id="save", 
# style="color:red").

def make_tag(**kwargs):
    return kwargs

make_tag(**{"class": "btn", "id": "save", "style": "color:red"})
```

```python
# company == Positional argument , *args : **kwargs
def show_elements(Company , *products):

    print("App name is " , Company)
    for i in products:
        print(i)
```
