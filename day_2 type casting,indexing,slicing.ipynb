# indexing in string 

- Every item in a list has a position number called an index. Indexing allows us to access a single item from the list using its position.
- Positive indexing starts from 0 for the first item, 1 for the second item, and so on.
- Negative indexing starts from -1 for the last item, -2 for the second last item, and so on.
- Negative indexing is useful when we want to access items from the end of the list without knowing its exact length.

- note string calculate space also while counting 


```python
x = 'i am software engineer'

print(x[-19])
print(x[6])
```

```python
cities = ["pune" , "banglore","nagpur","kochin", "latur","kolhapur"]
cities
print(cities [ 0 : 5])
```

```python
# indexing starts from 0 
# from ending it from -1

print(cities [0])
print(cities[0 : -1])
print(cities[3: -2])

#iwant last city 
print(cities[-1])

#i want blr and latur add +1 element
print(cities[4])
```

# Slicing
Slicing allows us to access a range or a portion of the list instead of a single item. The general syntax is:

list_name[start : stop : step]


```python
# Slicing in STring
x = 'i am software engineer'

print(x[5:13])

print(x[5 : ])

print(x[ : 5])

print(x[ : ])
```

```python
print(x[0: :3],) # prints every 3rd letter str also count space
```

```python
name = 'Shubham'
print(name[ : : -1]) # reverse string

print(x[14 : ])

print(x[-1 : -8])
```
