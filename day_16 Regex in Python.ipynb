# What is Regex?
- Regex (Regular Expression) is a way to search, match, or manipulate text using patterns.
Think of it like a "smart search tool" that can find words, numbers, or specific formats inside strings.

# Common Regex Notations in Python
- \d → Matches any digit (0–9)
- \D → Matches any non-digit character
- \w → Matches any word character (letters, digits, underscore)
- \W → Matches any non-word character
- \s → Matches any whitespace (space, tab, newline)
- \S → Matches any non-whitespace character
- . → Matches any single character (except newline)
- ^ → Matches the start of a string
- $ → Matches the end of a string
- → Matches one or more repetitions
- ***** → Matches zero or more repetitions
- ? → Matches zero or one occurrence
- {n} → Matches exactly n repetitions
- {n,m} → Matches between n and m repetitions
- | → OR operator (choice between patterns)
- [] → Matches any single character inside the brackets
- [^] → Matches any character not inside the brackets

- Regular expressions work by matching a sequence of characters against a defined pattern. 
- The Python `re` module provides functions to perform these operations. When a regex engine processes a pattern, it attempts to find an occurrence of that pattern within a target string. If a match is found, the engine returns information about the match; otherwise, it indicates that no match was found.

### Rules for regex

1.  **Specificity**: Be as specific as possible to avoid unintended matches.
2.  **Escaping**: Use a backslash (`\`) to escape metacharacters if you want to match them literally (e.g., `\.` to match a literal dot).
3.  **Raw Strings**: In Python, use raw strings (prefix with `r`, e.g., `r'\d+'`) for regular expressions to avoid issues with backslash escapes being interpreted by Python itself.
4.  **Greediness**: Quantifiers (`*`, `+`, `?`, `{}`) are by default 'greedy', meaning they will match the longest possible string. Add `?` after them to make them 'non-greedy' (e.g., `*?`).
5.  **Anchors**: Use `^` and `$` to anchor patterns to the beginning or end of a string (or lines in multiline mode) for precise matches.

### Demonstration : regex on a paragraph

- Here is a sample paragraph to demonstrate various regex operations. This paragraph contains several words, numbers, and punctuation marks. You can search for specific patterns like email addresses, phone numbers (though none are explicitly here, we can simulate), or just common words. The goal is to illustrate how flexible and powerful regular expressions are for extracting information from text data. Remember, Python's `re` module is your friend for these tasks.

```python
import re

# 1. Plus (+) → Matches 1 or more repetitions
# Matches 'a' followed by one or more 'b's
print("1. Plus (+):", re.findall(r"ab+", "a ab abb ac")) 
# Output: ['ab', 'abb']

# 2. Asterisk (*) → Matches 0 or more repetitions
# Matches 'a' followed by zero or more 'b's
print("2. Asterisk (*):", re.findall(r"ab*", "a ab abb ac")) 
# Output: ['a', 'ab', 'abb', 'a']

# 3. Question Mark (?) → Matches 0 or 1 occurrence
# Matches 'a' followed by an optional 'b'
print("3. Question Mark (?):", re.findall(r"ab?", "a ab abb ac")) 
# Output: ['a', 'ab', 'ab', 'a']

# 4. Exact Braces {n} → Matches exactly n repetitions
# Matches exactly 3 'a's
print("4. Exact {n}:", re.findall(r"a{3}", "aa aaa aaaa")) 
# Output: ['aaa', 'aaa']

# 5. Range Braces {n,m} → Matches between n and m repetitions
# Matches between 2 and 3 'a's
print("5. Range {n,m}:", re.findall(r"a{2,3}", "a aa aaa aaaa")) 
# Output: ['aa', 'aaa', 'aaa']

# 6. Pipe (|) → OR operator (choice between patterns)
# Matches either 'cat' or 'dog'
print("6. Pipe (|):", re.findall(r"cat|dog", "the cat and the dog")) 
# Output: ['cat', 'dog']

# 7. Brackets [] → Matches any single character inside
# Matches 'h', then 'a' or 'i', then 't'
print("7. Brackets []:", re.findall(r"h[ai]t", "hat hit hot")) 
# Output: ['hat', 'hit']

# 8. Negated Brackets [^] → Matches any character NOT inside
# Matches 'h', then any character EXCEPT 'o', then 't'
print("8. Negated [^]:", re.findall(r"h[^o]t", "hat hit hot")) 
# Output: ['hat', 'hit']
```
