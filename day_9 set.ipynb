#   SETS
     A set is an unordered collection of unique elements.

## Characteristics of Sets
- **Unordered** → No fixed position of elements.
- **Unique elements** → Duplicates are automatically removed.
- **Mutable container** → You can add or remove elements.
- **Immutable members** → Elements must be immutable types.

````python
# find the max number without using max()
nums = [56,67,78,456,43,56,76,8,56]

# 1. Initialize a variable with the first element of the list
max_num = nums[0]

# 2. Loop through each number in the list
for num in nums:
    # 3. If the current number is greater than our tracked max, update it
    if num > max_num:
        max_num = num

# 4. Print the final maximum number
print("The maximum number is:", max_num)





````
