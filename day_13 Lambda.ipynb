## lAMBDA FUNCTION

What is a Lambda Function?
- A lambda function is a small, anonymous & temporary function in Python.
-  defined using keyword : lambda instead of def.
- Unlike normal functions, a lambda function has no name and is usually used for short, simple operations.

- write in single line
- when we wanted a quick code
- Commonly used with functions like map(), filter(), and sorted().
- Helps keep code concise and readable.



## Syntax:

(lambda arguments: expression)

- arguments: Input values (like parameters in a normal function).
-  expression: A single statement that is evaluated and returned.


## Key Characteristics
- Anonymous: No function name is required.
- Single Expression: Can only contain one expression (no multiple statements).
- Inline Use: Often used where functions are needed temporarily.

```python
# traditinal way
def add(a, b ):
    return a + b 
```

```python
# 1. Write a lambda function to add two numbers.
add = lambda a,b : a+b
add(4,5)  #function calling
```

```python
# 2. Write a lambda function to square a number.

sq= lambda a : a*a
sq(2)
```

```python
#3. Write a lambda function to find the cube of a number.
cube = lambda a : a*a*a
cube(3)
```

```python
#4. Write a lambda function to check if a number is even.

no = lambda t : "even" if t % 2 == 0 else "odd"
print(no(5))
```

```python
# 5. Write a lambda function to find the Largest Numbers

v = lambda a,b :a if a > b else b
print("the largest no is :",v(100,9))
print("the largest no is :",v(1,9))
```
