### while Loop – Example 2: Countdown from 5 to 1

```python
countdown = []
n = 5
while n > 0:
    countdown.append(n)
    n = n - 1
countdown
```

### while Loop – Example 4: Collecting Even Numbers Below 20

```python
evens = []
i = 0
while i < 20:
    evens.append(i)
    i = i + 2
evens
```

```python
# simulating 3 password attempts

password_attempts = []
attempt = 1
while attempt <= 3:
    password_attempts.append(attempt)
    attempt = attempt + 1
password_attempts
```
