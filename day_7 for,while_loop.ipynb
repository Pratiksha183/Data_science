# while loop : 
- used when we don't know how many iterations will be done

```python
# 1 print 1 to 10 using while loop
i = 1                           # initialization start pt
while i<=10 :                   # end of loop
    print(i,end=" ")                    # print no
    i = i+1                     # increment
```

```python
#example 2:  printing no from 10 to 1 using reverse while loop
i = 10                          # initialization start pt
while i>=1 :                   # end of loop
    print(i)                    # print no
    i -= 1  
```

```python
# 3. print table of the no. given by user
n = int(input("enter your no :"))

i = 1
while i <=10:
    print(i*n)
    i += 1
```

```python
# Example 4: Countdown from 5 to 1
countdown = []
i = 5
while i > 0:
    countdown.append(i)
    i = i - 1
countdown
```
