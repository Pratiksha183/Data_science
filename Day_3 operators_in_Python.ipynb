## Types of Operators
- 1.Arithmetic Operators 
- 2.comparision / relational 
- 3.logical
- 4.assignment


```python
# Arithmetic operators :Perform mathematical calculations.

Arithmetic Examples:

+ (Addition) → 5 + 3 = 8
- (Subtraction) → 10 - 4 = 6
* (Multiplication) → 4 * 3 = 12
/ (Division) → 7 / 2 = 3.5
// (Floor Division) → 7 // 2 = 3  # 3 if u use // u wont get answer in float type value
% (Modulus) → 7 % 2 = 1   # 1 moduls [remender]
** (Exponentiation) → 2 ** 3 = 8  #power 2^3 = 8
```

```python
Comparison Operators/relational :

Compare values and return True or False.
Examples:

< (Less than) → 3 < 5 → True
> (Greater than) → 7 > 2 → True
<= (Less than or equal to) → 4 <= 6 → True
>= (Greater than or equal to) → 5 >= 5 → True
== (Equal to) → 3 == 3 → True
!= (Not equal to) → 3 != 5 → True
```

```python
Logical Operators :

Combine conditional statements.
Examples:

and → True if both conditions are true
or → True if at least one condition is true
not → Negates the condition
```

```python
a = 10
b = 3

print(a + b)  
print(a - b)   
print(a * b)   
print(a / b)   
print(a % b)   # 1 moduls [remender]


# arithmatic operators 

"""
 + → addition
- → subtraction
* → multiplication
/ → division
% → modulus (remainder)
** → exponent power 2^3 = 8
// → floor division 

"""



```
