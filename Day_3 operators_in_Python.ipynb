#### 2.Comaprision Operator / relational  ( == , != , > , < , > ) :

- Compare values and return True or False.
- < (Less than) → 3 < 5 → True
- > (Greater than) → 7 > 2 → True
- <= (Less than or equal to) → 4 <= 6 → True
- >= (Greater than or equal to) → 5 >= 5 → True
- == (Equal to) → 3 == 3 → True
- != (Not equal to) → 3 != 5 → True

```python
x = 50
y = 30

print(x == y)
print(x != y)
print(x > y)
print(x < y)
```

```python
x = 50
y = 30

print(x == y)
print(x != y)
print(x > y)
print(x < y)
```

### 3.  Logical Operators :( AND , OR , NOT )

- and -Returns True only if both conditions are true.
- OR -Returns True if at least one condition is true.
- NOR-Reverses the result of a condition.

```python
x = 50
y = 30

print( x > 10 and y > 100)

print( x > 100 or y > 100)

print( not x > 10)
```

# ** 5.identity Operator :- ( is , is not ) **

- compares location of 2 variables nd if both same returns true

```python
a = 10
b = 10

print('A Location →', id(a))

print('B Location is →', id(b))
```

```python
print(a is b)
print(a is not b)
```
