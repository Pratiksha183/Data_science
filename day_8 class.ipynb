```python
marks = [45,56,78,52,57,78,1,4]
```

```python
for i in marks:
    if i < 35:
        print(i , "you failed you need a feedback")
      
        break  # terminate at 1 won't execute for 4
    
    print(i)
```

```python
for i in marks:
    if i < 35:
        print(i , "you failed you need a feedback")

        continue   # allow to execute for 4 also
      
        break
    
    print(i)
```

```python
# Funny continue example with pets
pets = ["dog", "cat", "snake", "parrot"]

for pet in pets:
    if pet == "snake":
        print("Nope! Skip the scary:", pet)
        continue
    print("I like this pet:", pet)
```

```python
# Marks list with continue
marks = [85, 30, 75, 40, 25, 90]

for score in marks:
    if score < 40:   # skip failed marks
        print("Skipping fail:", score)
        continue
    print("Pass:", score)
```
