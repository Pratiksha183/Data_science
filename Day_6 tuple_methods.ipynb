# tuple :
- A tuple is an ordered collection of elements in Python

-  immutable : meaning their contents cannot be changed after creation. 
- ordered :the items in a tuple have a defined order, and that order will not change
- Immutable: Once created, the elements of a tuple cannot be modified.
- Allow Duplicates: Tuples can contain duplicate values.
- Indexed: Tuple items are indexed, with the first item having index [0].
- Heterogeneous: A tuple can store elements of different data types
(integers, strings,   floats, lists, etc.).
