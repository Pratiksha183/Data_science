```python
# Import the datetime module and print the current date and time.

import datetime as d
import pytz

z = pytz.timezone("Asia/Kolkata")

da = d.datetime.now(z)
print(da)

```

```python
# Print only the current time.

import datetime as d
import pytz

z = pytz.timezone("Asia/Kolkata")

da = d.datetime.now(z).time()
print(da)
```

```python
# Print only the current date.

import datetime as d

dt = d.date.today()
print(dt)
```

```python
# Create a date object manually 

import datetime as d

D = d.datetime(2026,10,14,4,20,30)
print(D)
```

```python
print(D.strftime("%A")) # Day
print(D.strftime("%a")) # day in short 
print(D.strftime("%b")) # Month in short
print(D.strftime("%B")) # Month
print(D.strftime("%m")) # Month in number
print(D.strftime("%M")) # Minutes
print(D.strftime("%H")) # Hours
```

```python
#Extract year, month, and day from the current date.

from datetime import datetime

now = datetime.now()
print(now.year)
print(now.month)
print(now.day)
```

```python
# Extract hour, minute, and second.

from datetime import datetime

now = datetime.now()

print(now.hour)
print(now.minute)
print(now.second)
```

```python
# Print the current weekday.

from datetime import datetime

today = datetime.now()
print(today.strftime("%A"))
```

```python
# Convert current date into DD-MM-YYYY format.

from datetime import datetime

today = datetime.now()
print(today.strftime("%d/%m/%Y"))
```

```python
# Convert current date into YYYY/MM/DD format.

from datetime import datetime

today = datetime.now()
print(today.strftime("%Y/%m/%d"))
```

```python
# Print current time in 12-hour format.

from datetime import datetime

now = datetime.now()
print(now.strftime("%I:%M:%S %p"))
```

```python
# Add 10 days to the current date.

from datetime import datetime, timedelta

future = datetime.now() + timedelta(days=10)
print(future)
```

```python
# Subtract 5 days from the current date.

from datetime import datetime, timedelta

past = datetime.now() - timedelta(days=5)
print(past)
```

```python
# Find the difference between two dates.

from datetime import date

d1 = date(2025, 1, 1)
d2 = date(2025, 1, 20)
print(d2 - d1)
```

```python
# Find your age using birth year.

from datetime import date

birth_year = 2003
current_year = date.today().year
age = current_year - birth_year
print(age)
```
