# PYTHON DICTIONARY {} 

Dictionary is a built-in data type used to store key-value pairs.
It is unordered, mutable, and indexed by keys.

- it doesn't allow duplicate keys {value can be duplicate}

```python
user = dict(name="John", age=36, country="Norway",city="Norway")
print(user)
```

```python
user.items()
```

Keys → Unique identifiers (immutable types like strings, numbers, tuples). = name
Values → Data associated with keys (can be any type) =Rock
